### 1.14 Interpretation of Institutions Missing Between Quarters

The comparison between Q1 and Q2 identified 45 institutions present only in Q1 and 4 institutions present only in Q2.

The Q1-only records showed three end-effective date patterns:
- 38 records had end-effective dates on or before June 30, 2026.
- 2 records had end-effective dates after June 30, 2026.
- 5 records contained the value `99991231`, which appears to represent an open-ended date marker.

Additionally, only 4 of the 45 Q1-only records contained an `ACTIVE` value, and all four were marked as active.

For the Q2-only records, 23 columns were completely missing across all four records.

These observations indicate differences in institution records between reporting quarters. However, the available information is insufficient to classify all unmatched institutions as closures, failures, or newly established banks.

Further verification is required before assigning explanations to these differences.

```python
# Convert end-effective dates into datetime values

end_dates = pd.to_datetime(
    q1_only["ENDEFYMD"].astype("string"),
    format="%Y%m%d",
    errors="coerce"
)

q2_end_date = pd.Timestamp("2026-06-30")
open_ended_date = pd.Timestamp("9999-12-31")

date_categories = pd.Series(
    "Missing or invalid",
    index=q1_only.index,
    dtype="object"
)

# Identify open-ended records
is_open_ended = end_dates == open_ended_date

date_categories.loc[is_open_ended] = "Open-ended"

# Dates on or before Q2 reporting date
date_categories.loc[
    end_dates.notna() &
    ~is_open_ended &
    (end_dates <= q2_end_date)
] = "On or before Q2 reporting date"

# Dates after Q2 reporting date, excluding open-ended records
date_categories.loc[
    end_dates.notna() &
    ~is_open_ended &
    (end_dates > q2_end_date)
] = "After Q2 reporting date"

display(date_categories.value_counts().to_frame("Count"))
```

### Balance-Sheet Consistency Check

A balance-sheet consistency check is performed by comparing total assets with the sum of total liabilities and equity.

The difference between these values is examined to identify records that may require further investigation. Records with missing equity are excluded from this particular check.

This is an initial consistency test, not a definitive audit of the reported financial statements.

```python
# Check the accounting relationship between assets, liabilities and equity

balance_check = q2.dropna(
    subset=["ASSET", "LIAB", "EQ"]
).copy()

balance_check["BALANCE_DIFFERENCE"] = (
    balance_check["ASSET"] -
    balance_check["LIAB"] -
    balance_check["EQ"]
)

print("Records checked:", len(balance_check))

print(
    "Exact matches:",
    (balance_check["BALANCE_DIFFERENCE"] == 0).sum()
)

print(
    "Non-zero differences:",
    (balance_check["BALANCE_DIFFERENCE"] != 0).sum()
)

print(
    "Mean absolute difference:",
    balance_check["BALANCE_DIFFERENCE"].abs().mean()
)

print(
    "Maximum absolute difference:",
    balance_check["BALANCE_DIFFERENCE"].abs().max()
)

print("\n10 largest absolute differences:")

display(
    balance_check.assign(
        ABS_DIFFERENCE=balance_check["BALANCE_DIFFERENCE"].abs()
    ).nlargest(10, "ABS_DIFFERENCE")[
        [
            "CERT", "NAME", "BKCLASS",
            "ASSET", "LIAB", "EQ",
            "BALANCE_DIFFERENCE"
        ]
    ].to_string(index=False)
)
```

### Investigation of Balance-Sheet Differences

A total of 94 institutions showed non-zero differences when comparing total assets with the sum of total liabilities and equity.

These records are examined to understand the magnitude and distribution of the differences. This investigation will help determine whether the differences are small relative to institution size or require further attention.

```python
# Create a copy of records with non-zero balance-sheet differences

nonzero_balance_records = balance_check[
    balance_check["BALANCE_DIFFERENCE"] != 0
].copy()

# Calculate the difference relative to total assets
nonzero_balance_records["DIFFERENCE_PERCENTAGE"] = (
    nonzero_balance_records["BALANCE_DIFFERENCE"].abs()
    / nonzero_balance_records["ASSET"].abs()
    * 100
)

# Inspect the largest relative differences
display(
    nonzero_balance_records.sort_values(
        "DIFFERENCE_PERCENTAGE",
        ascending=False
    )[
        [
            "CERT", "NAME", "BKCLASS",
            "ASSET", "LIAB", "EQ",
            "BALANCE_DIFFERENCE",
            "DIFFERENCE_PERCENTAGE"
        ]
    ].head(15).to_string(index=False)
)
```

### Distribution of Relative Balance-Sheet Differences

The relative differences are summarized to understand how frequently balance-sheet differences occur at different magnitudes.

This provides an overview of the issue across institutions instead of focusing only on the largest individual observations.

```python
# Summarize relative balance-sheet differences

relative_diff = nonzero_balance_records["DIFFERENCE_PERCENTAGE"]

print("Total records with non-zero differences:", len(relative_diff))

print("\nRelative Difference Summary:")
display(relative_diff.describe().round(4))

print("\nRecords above selected thresholds:")

print("Above 0.1%:", (relative_diff > 0.1).sum())
print("Above 0.5%:", (relative_diff > 0.5).sum())
print("Above 1%:", (relative_diff > 1).sum())

print("\nDifference direction:")
print(
    "Assets greater than liabilities + equity:",
    (nonzero_balance_records["BALANCE_DIFFERENCE"] > 0).sum()
)
print(
    "Assets less than liabilities + equity:",
    (nonzero_balance_records["BALANCE_DIFFERENCE"] < 0).sum()
)
```

### Interpretation of Balance-Sheet Consistency Results

Out of 4,238 institutions with available asset, liability, and equity values, 4,144 records satisfied the exact relationship between total assets, total liabilities, and equity. The remaining 94 records showed non-zero differences.

Most differences were small relative to total assets. The median relative difference was approximately 0.0119%, while the mean was 0.1128%.

Twenty records had differences exceeding 0.1%, six exceeded 0.5%, and two exceeded 1%.

In 92 records, total assets exceeded the sum of liabilities and equity. In the remaining two records, the sum of liabilities and equity exceeded total assets.

These observations are retained for further investigation. They are not automatically classified as errors because reporting conventions, rounding, and field definitions must be verified first.

```python
# Inspect records with relative differences above 1%

print("Records with relative differences above 1%:")

display(
    nonzero_balance_records[
        nonzero_balance_records["DIFFERENCE_PERCENTAGE"] > 1
    ][
        [
            "CERT", "NAME", "BKCLASS",
            "ASSET", "LIAB", "EQ",
            "BALANCE_DIFFERENCE",
            "DIFFERENCE_PERCENTAGE"
        ]
    ].to_string(index=False)
)

# Inspect records where liabilities + equity exceed assets

print("\nRecords where liabilities + equity exceed assets:")

display(
    nonzero_balance_records[
        nonzero_balance_records["BALANCE_DIFFERENCE"] < 0
    ][
        [
            "CERT", "NAME", "BKCLASS",
            "ASSET", "LIAB", "EQ",
            "BALANCE_DIFFERENCE",
            "DIFFERENCE_PERCENTAGE"
        ]
    ].to_string(index=False)
)
```

## 21. NIM and Bank Size Relationship

This section explores the relationship between reported NIM values and total bank assets. The analysis helps determine whether the large numerical values observed in NIM are associated with larger institutions.

```python
nim_asset_data = q2[["ASSET", "NIM"]].dropna()

print("Correlation between ASSET and NIM:")
print(nim_asset_data["ASSET"].corr(nim_asset_data["NIM"]))

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))
plt.scatter(nim_asset_data["ASSET"], nim_asset_data["NIM"], alpha=0.4)
plt.xlabel("Total Assets")
plt.ylabel("Reported NIM")
plt.title("Relationship Between Bank Assets and NIM")
plt.show()
```

### Interpretation

The analysis produced a Pearson correlation coefficient of 0.9653 between total assets and reported NIM, indicating a very strong positive linear association.

Larger institutions generally report higher numerical NIM values. However, the relationship may be strongly influenced by the presence of a few very large banks.

The exact meaning and measurement unit of NIM have not yet been verified through the FDIC dictionary. Therefore, this result is treated as an exploratory numerical relationship rather than evidence about conventional net interest margin performance.

```python
nim_asset_data = q2[["ASSET", "NIM"]].dropna().copy()

asset_cutoff = nim_asset_data["ASSET"].quantile(0.99)

filtered_data = nim_asset_data[
    nim_asset_data["ASSET"] <= asset_cutoff
]

print("Original correlation:",
      nim_asset_data["ASSET"].corr(nim_asset_data["NIM"]))

print("99th percentile asset cutoff:", asset_cutoff)

print("Correlation after excluding largest 1%:",
      filtered_data["ASSET"].corr(filtered_data["NIM"]))

print("Original records:", len(nim_asset_data))
print("Filtered records:", len(filtered_data))
```

### Influence of Large Institutions

To examine whether the relationship between total assets and reported NIM was dominated by the largest institutions, the Pearson correlation was recalculated after excluding the top 1% of institutions by total assets.

The original correlation was 0.9653. After excluding 43 institutions above the 99th percentile of total assets, the correlation decreased to 0.8977.

The relationship remained strongly positive, suggesting that the association is not solely driven by the largest institutions. However, the exact interpretation of NIM remains subject to verification of its official definition and measurement unit.

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    filtered_data["ASSET"],
    filtered_data["NIM"],
    alpha=0.4
)

plt.xlabel("Total Assets")
plt.ylabel("Reported NIM")
plt.title("Assets vs NIM (Excluding Largest 1% of Banks)")
plt.grid(True, alpha=0.3)

plt.show()
```
